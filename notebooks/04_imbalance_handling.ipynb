```python
import pandas as pd

from imblearn.over_sampling import SMOTE
```

```python
df = pd.read_csv(r'C:\Users\joshn\OneDrive\Documents\credit-card-fraud-detection\data\raw\creditcard.cs')
```

```python
X = df.drop('Class', axis=1)

y = df['Class']
```

```python
smote = SMOTE(random_state=42)

X_resampled, y_resampled = smote.fit_resample(X, y)
```
